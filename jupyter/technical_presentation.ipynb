# Pinyon Detection Technical Presentation

## Weighted Centers and Cluster splitting
Weighted centers were originally used early on in the project to make sure labels would select the closest tree, regardless of grass pulling the geographic center of a cluster away from it. Later on, they were used to identify clusters with multiple trees and split them. Splitting clusters this way worked very well, but didn't always trigger when needed. Any resulting clusters went back into the array to then be filtered and labeled. 

## Filtering Clusters
After any new clusters are formed, they got filtered before labeling. Prospective clusters get filtered by their radius, point density, greenness, and height before labeling.
